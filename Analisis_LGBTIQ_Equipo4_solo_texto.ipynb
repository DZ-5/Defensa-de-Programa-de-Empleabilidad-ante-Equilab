# Análisis de la población LGBTIQ+ — Equilab
### Equipo 4 · Brecha laboral de personas LGBTIQ+ · GEIH 2025

**Contexto:** Equilab tiene presupuesto para financiar un único programa de
empleabilidad en 2027. Este documento reúne, en lenguaje sencillo, la evidencia
recolectada con la Gran Encuesta Integrada de Hogares (GEIH) para responder si la
población LGBTIQ+ debería ser la población priorizada, y en qué dimensión del mercado
laboral tendría más sentido intervenir.

No incluye código ni consultas SQL: es la versión de lectura, pensada para acompañar
la presentación ante el comité.

## Punto 1. ¿Cuántas personas son?

**Qué se mide:** el tamaño de la población LGBTIQ+ en 2025 -- cuántas personas
representa esa población a nivel nacional, y cuántas de ellas están ocupadas.

| Personas | Registros | Ocupados |
|---|---|---|
| 472.246 | 6.343 | 348.229 |



*Gráfica 1. Magnitud de la población LGBTIQ+, 2025.*

**Análisis:** en 2025, la población LGBTIQ+ representada en la GEIH corresponde a
aproximadamente **472.246 personas**, respaldadas por **6.343 registros** de la
encuesta. De esta población, aproximadamente **348.229 personas se encontraban
ocupadas**.

El número de registros es superior a 100, por lo que no se presenta una alerta de
muestra pequeña según el criterio establecido en el taller. Es importante diferenciar
entre los registros de la encuesta y la población estimada: los **6.343 registros**
representan aproximadamente **472.246 personas** mediante el factor de expansión
`fex`.

## Punto 2. ¿Cómo les va?

**Qué se mide:** se compara la población LGBTIQ+ ocupada con el resto de la población
ocupada en 2025, en tres indicadores laborales: informalidad, cotización a pensión, y
existencia de contrato escrito.

| Grupo | Registros | Informalidad | Cotiza a pensión | Contrato escrito |
|---|---|---|---|---|
| Resto | 350.676 | 55,1 % | 43,4 % | 40,0 % |
| LGBTIQ+ | 4.499 | 37,3 % | 62,0 % | 57,9 % |



*Gráfica 2. Condiciones laborales, LGBTIQ+ vs. Resto (2025).*

**Análisis:** en 2025, la población LGBTIQ+ ocupada presenta una **informalidad del
37,3 %**, mientras que el resto de la población ocupada presenta una informalidad del
**55,1 %**. Esto representa una diferencia de **17,8 puntos porcentuales** a favor de
la población LGBTIQ+.

En cuanto a la cotización a pensión, el **62,0 %** de la población LGBTIQ+ ocupada
cotiza, frente al **43,4 %** del resto, una diferencia de **18,6 puntos porcentuales**.

Respecto al contrato escrito, el **57,9 %** de la población LGBTIQ+ cuenta con este
tipo de contrato, frente al **40,0 %** del resto, una diferencia de **17,9 puntos
porcentuales**.

Por lo tanto, en los tres indicadores analizados, la población LGBTIQ+ presenta
resultados más favorables que el resto de la población ocupada.

Sin embargo, estos resultados muestran **diferencias entre los grupos, no una
relación causal**. Es decir, con esta información todavía no podemos afirmar por qué
existen estas diferencias -- por eso el siguiente punto revisa si los dos grupos son
comparables entre sí.

## Punto 3. ¿La brecha es real o es composición?

**Qué se mide:** antes de interpretar la diferencia de informalidad, es necesario
comprobar si los dos grupos son comparables de entrada. Se revisan tres
características de composición: zona, educación superior y edad promedio.

| Grupo | Registros | % Urbano | % Educación superior | Edad promedio |
|---|---|---|---|---|
| Resto | 350.676 | 79,8 % | 33,4 % | 41,3 años |
| LGBTIQ+ | 4.499 | 92,5 % | 60,0 % | 34,4 años |



*Gráfica 3. Composición de los dos grupos (2025).*

**Análisis:** los resultados muestran que los dos grupos **no tienen la misma
composición**.

La población LGBTIQ+ es más urbana: **92,5 %** vive en zona urbana, frente al **79,8 %**
del resto.

También presenta una mayor proporción de educación superior: **60,0 %** frente al
**33,4 %** del resto, una diferencia de **26,6 puntos porcentuales**.

Además, la edad promedio de la población LGBTIQ+ es de **34,4 años**, mientras que la
del resto es de **41,3 años**, es decir, aproximadamente **6,9 años menos**.

Estas diferencias son importantes porque pueden influir en los resultados laborales.
Por esta razón, no es suficiente quedarse solamente con la comparación inicial de
informalidad. La pregunta pasa a ser: **entre personas con características similares,
¿la brecha de informalidad continúa?**

### Punto 3.1. Control por zona

Para comprobar si la diferencia de informalidad se debe únicamente a la composición
territorial, se vuelve a realizar la comparación **dentro de cada zona**: LGBTIQ+
rural vs. resto rural, y LGBTIQ+ urbano vs. resto urbano.

| Zona | Grupo | Registros | Informalidad |
|---|---|---|---|
| Rural | LGBTIQ+ | 240 | 74,1 % |
| Rural | Resto | 42.274 | 83,1 % |
| Urbano | LGBTIQ+ | 4.259 | 34,3 % |
| Urbano | Resto | 308.402 | 48,0 % |



*Gráfica 4. Informalidad controlando por zona, 2025.*

**Análisis:** al controlar la comparación por zona, la diferencia de informalidad
**se mantiene tanto en el área rural como en la urbana**.

En la zona rural, la informalidad de la población LGBTIQ+ es de **74,1 %**, frente al
**83,1 %** del resto -- una diferencia de aproximadamente **8,9 puntos porcentuales**.

En la zona urbana, la población LGBTIQ+ presenta una informalidad del **34,3 %**,
mientras que el resto presenta **48,0 %**, una diferencia de **13,7 puntos
porcentuales**.

La comparación inicial mostraba una diferencia de **17,8 puntos porcentuales**. Después
de controlar por zona, la diferencia se reduce, pero **no desaparece**.

Esto significa que la composición territorial explica **una parte** de la diferencia
inicial, pero no toda. La población LGBTIQ+ continúa presentando una menor
informalidad incluso comparándose con el resto de personas dentro de la misma zona.
No sería correcto decir que la brecha desaparece por composición, ni que pertenecer a
la población LGBTIQ+ *causa* una menor informalidad. Lo que sí podemos afirmar es que
**la diferencia permanece después de controlar por zona**.

## Punto 4. ¿Cómo cambió entre 2021 y 2025?

**Qué se mide:** la evolución de la informalidad de la población LGBTIQ+ entre 2021 y
2025, con el porcentaje de informalidad y el número de registros de cada año.

| Año | Registros | Informalidad |
|---|---|---|
| 2021 | 6.324 | 50,0 % |
| 2022 | 4.698 | 45,1 % |
| 2023 | 4.497 | 42,7 % |
| 2024 | 4.277 | 39,9 % |
| 2025 | 4.499 | 37,3 % |



*Gráfica 5. Evolución de la informalidad LGBTIQ+, 2021-2025.*

**Análisis:** entre 2021 y 2025 se observa una **tendencia descendente de la
informalidad** en la población LGBTIQ+. En 2021 la informalidad era del **50,0 %**,
mientras que en 2025 fue del **37,3 %** -- una reducción de **12,7 puntos
porcentuales**, progresiva año a año.

Todos los años cuentan con más de 100 registros, por lo que no existe una alerta de
muestra pequeña según el criterio del taller.

Estos resultados permiten identificar una disminución de la informalidad durante el
periodo analizado. Sin embargo, **los datos no permiten determinar por sí solos las
causas de esta reducción**.

## Punto 5. ¿Dónde es peor?

**Qué se mide:** un análisis territorial para identificar los departamentos con
mayores niveles de informalidad entre la población LGBTIQ+ ocupada en 2025. Para
evitar conclusiones basadas en muestras pequeñas, se incluyen únicamente
departamentos con mínimo 100 registros.

| Departamento | Registros | Informalidad |
|---|---|---|
| Nariño | 299 | 75,6 % |
| Norte de Santander | 147 | 63,2 % |
| Cauca | 122 | 62,9 % |
| Tolima | 118 | 62,5 % |
| Atlántico | 158 | 55,5 % |
| Santander | 224 | 49,1 % |



*Gráfica 6. Departamentos con mayor informalidad LGBTIQ+, 2025.*

**Análisis:** en 2025, **Nariño** presenta el mayor nivel de informalidad de la
población LGBTIQ+ entre los departamentos que cuentan con al menos 100 registros, con
**75,6 % y 299 registros**. Le siguen Norte de Santander con **63,2 %**, Cauca con
**62,9 %** y Tolima con **62,5 %**.

Estos resultados muestran que existen diferencias territoriales importantes en la
informalidad de la población LGBTIQ+. Es importante aclarar que no se puede afirmar
que Nariño tenga la mayor informalidad de todo el país sin condición: la afirmación
correcta es que **Nariño presenta el mayor nivel entre los departamentos que cumplen
el requisito mínimo de 100 registros**. Los departamentos que no aparecen en la tabla
fueron excluidos por tener menos de 100 registros, por lo que no deben usarse para
sacar conclusiones fuertes.

## Punto 6. Recomendación

Con la evidencia reunida en los cinco puntos anteriores, la conclusión que surge de
los datos es distinta a la que se podría anticipar antes de mirarlos: **en los tres
indicadores analizados (informalidad, pensión y contrato escrito), la población
LGBTIQ+ ocupada no muestra una desventaja frente al resto de la población -- muestra
una ventaja**, de entre 17,8 y 18,6 puntos porcentuales según el indicador.

Parte de esa ventaja se explica por la composición del grupo: la población LGBTIQ+
encuestada es más urbana, más educada y más joven que el resto, tres características
que por sí solas se asocian a menor informalidad. Pero al controlar por zona (Punto
3.1), la ventaja se reduce -- de 17,8 a 8,9 puntos en zona rural y a 13,7 puntos en
zona urbana -- sin desaparecer en ningún caso. Es decir, la composición explica una
parte de la diferencia, no toda.

La informalidad de la población LGBTIQ+ además viene bajando de forma sostenida (de
50,0 % en 2021 a 37,3 % en 2025), y la brecha territorial muestra que el problema no
es uniforme: se concentra con más fuerza en departamentos como Nariño, Norte de
Santander, Cauca y Tolima.

**Recomendación:** no sería adecuado diseñar el programa de Equilab bajo la premisa de
que la población LGBTIQ+ tiene, en general, peor calidad de empleo -- estos tres
indicadores muestran justamente lo contrario. En cambio, sí sería razonable enfocar
una intervención en los **territorios donde la informalidad LGBTIQ+ es más alta**, y
complementar este análisis con otras dimensiones del mercado laboral no cubiertas
aquí (por ejemplo, qué tan fácil es *conseguir* empleo, no solo qué tan buena es su
calidad una vez dentro), antes de decidir la focalización final del programa.

**Lo que no podemos afirmar con estos datos:**
1. Que la orientación sexual o identidad de género *cause* estas diferencias -- solo
   observamos asociación, en un dato transversal de un solo momento.
2. Que la situación sea igual para todos los subgrupos dentro de "LGBTIQ+" -- la GEIH
   identifica esta población de forma agregada, sin distinguir, por ejemplo, entre
   personas trans, gay, lesbianas o bisexuales.
3. Que la lectura territorial cubra todo el país -- varios departamentos quedaron
   fuera del análisis por no alcanzar el mínimo de 100 registros.
